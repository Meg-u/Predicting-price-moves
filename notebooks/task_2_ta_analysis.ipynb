Quantitative analysis using yfinance and TaLib

```python
import pandas as pd
import ta
import os

def process_stock(file_path, output_folder):
    # Load dataset
    df = pd.read_csv(file_path, parse_dates=['Date'])
    df = df.sort_values('Date')

    # Calculate indicators
    df['rsi'] = ta.momentum.RSIIndicator(close=df['Close']).rsi()
    macd = ta.trend.MACD(close=df['Close'])
    df['macd'] = macd.macd()
    df['macd_signal'] = macd.macd_signal()
    df['sma_50'] = ta.trend.SMAIndicator(close=df['Close'], window=50).sma_indicator()
    df['sma_200'] = ta.trend.SMAIndicator(close=df['Close'], window=200).sma_indicator()

    # Optional: Bollinger Bands
    bb = ta.volatility.BollingerBands(close=df['Close'], window=20, window_dev=2)
    df['bb_upper'] = bb.bollinger_hband()
    df['bb_lower'] = bb.bollinger_lband()

    # Save to new CSV
    os.makedirs(output_folder, exist_ok=True)
    symbol = os.path.basename(file_path).replace(".csv", "")
    df.to_csv(os.path.join(output_folder, f"{symbol}_indicators.csv"), index=False)
    print(f"{symbol} processed and saved.")


process_stock("../data/yfinance_data/AAPL_historical_data.csv", "processed_yfinance")

```

```python
symbols = ["AAPL", "AMZN", "GOOG", "META", "MSFT", "NVDA", "TSLA"]
base_path = "../data/yfinance_data"
output_folder = "processed_yfinance"

for symbol in symbols:
    file_path = f"{base_path}/{symbol}_historical_data.csv"
    process_stock(file_path, output_folder)


```

Visualize indicators for one company

```python
import pandas as pd
import matplotlib.pyplot as plt


aapl = pd.read_csv("../processed_yfinance/AAPL_processed.csv", parse_dates=['Date'])

fig, axs = plt.subplots(3, 1, figsize=(12, 10), sharex=True)

# Close price with SMA50
axs[0].plot(aapl['Date'], aapl['Close'], label='Close Price', color='blue')
axs[0].plot(aapl['Date'], aapl['sma_50'], label='SMA 50', color='orange')
axs[0].set_title("AAPL Close Price and SMA50")
axs[0].legend()

# RSI
axs[1].plot(aapl['Date'], aapl['rsi'], label='RSI', color='purple')
axs[1].axhline(70, color='red', linestyle='--', label='Overbought')
axs[1].axhline(30, color='green', linestyle='--', label='Oversold')
axs[1].set_title("RSI Indicator")
axs[1].legend()

# MACD
axs[2].plot(aapl['Date'], aapl['macd'], label='MACD', color='black')
axs[2].axhline(0, color='gray', linestyle='--')
axs[2].set_title("MACD Indicator")
axs[2].legend()

plt.tight_layout()
plt.show()
```
